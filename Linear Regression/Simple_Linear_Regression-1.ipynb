```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme()
```

```python
X = np.array([1,3,4,6,7])   # years of experience
y = np.array([15,35,45,65,75])  # salary in thosunds
```

```python
sns.scatterplot(x = X , y = y)
```

```python
def make_predication(X,y,w,b):       # w -> weights and b-> bais
  m = X.shape[0]
  pred_list = np.zeros((m,))

  for i in range(m):
    pred_list[i] = w * X[i] + b

  return pred_list
```

```python
predications = make_predication(X,y,0,0)
print(predications)
sns.scatterplot(x = X,y = y)
plt.plot(X, predications)
```

```python
predications = make_predication(X,y,w =5,b = 3)
print(predications)
sns.scatterplot(x = X,y = y)
plt.plot(X, predications)
```

```python
predications = make_predication(X,y,w=10,b =5)
print(predications)
sns.scatterplot(x = X,y = y)
plt.plot(X, predications,color='red')
plt.show()
```

# cost function

```python
def compute_cost(X,y,w,b):
  m = X.shape[0]         # numbers of datapoints
  cost = 0.00

  for i in range(m):
    predication = w * X[i] + b      # apply this
    error = predication - y[i]
    error_squred = error ** 2

    cost = cost + error_squred


  cost = cost / (2 * m)
  return cost

```

```python
w = 10.0
b = 5.0
predications = make_predication(X,y,w=w,b =b)
print(predications)
sns.scatterplot(x = X,y = y)
plt.plot(X, predications,color='red')
plt.title(f"Cost:{compute_cost(X,y,w=w,b=b)}, for b = {b}, and w ={w}")
```

```python
w = []
cost = []


for i in range(-100,100):
  cost_i = compute_cost(X,y,w=i,b=0)
  w.append(i)
  cost.append(cost_i)

plt.plot(w,cost)
plt.xlabel("weights - W")
plt.ylabel("Cost - J(w)")
```

# Gradient

```python
def calculate_gradient(X,y,w,b):
   m = X.shape[0]         # numbers of datapoints
   gradient = 0

   dj_dw = 0.0
   dj_db = 0.0

   for i in range(m):
    predication = w * X[i] + b      # apply this
    error = predication - y[i]

    dj_dw = dj_dw + (error - X[i])
    dj_db = dj_db + error

   dj_dw = dj_dw / m
   dj_db = dj_db / m

   return dj_dw , dj_db

```

```python
calculate_gradient(X,y,w=10,b=5)

```

# Apply gredient descent

```python
def gredient_descent(X,y,w_input,b_input,max_iter, alpha= 0.01):
  w = w_input
  b = b_input
  cost_memo = []
  iteration = []

  for i in range(max_iter):
    dj_dw , dj_db =  calculate_gradient(X,y,w,b)

    # update
    w = w - alpha * dj_dw   #in this protion i bascially performs {W(old) - 0.01(alpha value) * dj_dw(derivate of w)}
    b = b - alpha * dj_db

    cost = compute_cost(X,y,w,b)
    cost_memo.append(cost)
    iteration.append(i)

    if i % 100 == 0:
      print(f"iteration:{i}, w:{w:0.4f}, b:{b:0.4f}, dj_dw:{dj_dw:0.4f}, dj_db:{dj_db:0.4f}, cost:{cost:0.4f}")

  return w,b,cost_memo,iteration
```

```python
w_final,b_final,cost_memo,iter = gredient_descent(X,y,w_input=0,b_input=0,max_iter=10000, alpha= 0.01)
```

```python
print(w_final,b_final)

predications = make_predication(X,y,w=w_final,b=b_final)
print(predications)
sns.scatterplot(x = X,y = y)
plt.plot(X, predications,color='red')
plt.title(f"Cost:{compute_cost(X,y,w=w_final,b=b_final)}, for b = {b_final}, and w ={w_final}")
```
